# 01 - Exploration des données


```python
import os, sys
sys.path.insert(0, os.path.abspath("..")) 
os.chdir("..")                               

import matplotlib.pyplot as plt
from PIL import Image

from src.config import load_config
from src.data import find_data_dir, build_dataframe, prepare_dataframe, split_dataframe

cfg = load_config()
```

## Nombre d'images par classe

```python
data = build_dataframe(find_data_dir(cfg["data_root"]))
data["classe"].value_counts().sort_index().plot.barh(figsize=(8, 8))
plt.title("Nombre d'images par classe")
plt.show()
```

## Quelques images au hasard

```python
df, classes = prepare_dataframe(data, cfg["seed"], cfg["n_par_classe"])
exemples = df.sample(8, random_state=cfg["seed"])
plt.figure(figsize=(14, 7))
for k, (_, ligne) in enumerate(exemples.iterrows()):
    plt.subplot(2, 4, k + 1)
    plt.imshow(Image.open(ligne["chemin"]).convert("RGB"))
    plt.title(ligne["classe"].replace("___", "\n"), fontsize=8)
    plt.axis("off")
plt.subplots_adjust(top=0.88, hspace=0.4)
plt.show()
```

## Taille des jeux train / validation / test

```python
train_df, val_df, test_df = split_dataframe(df, cfg["seed"])
print("train :", len(train_df), "| validation :", len(val_df), "| test :", len(test_df))
```
