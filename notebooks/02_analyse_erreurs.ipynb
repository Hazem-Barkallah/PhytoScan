# 02 - Analyse des erreurs


```python
import os, sys
sys.path.insert(0, os.path.abspath(".."))   # pour importer le package src/ depuis notebooks/
os.chdir("..")                               # chemins relatifs à la racine du repo

import matplotlib.pyplot as plt
from PIL import Image

from src.config import load_config
from src.data import find_data_dir, build_dataframe, prepare_dataframe, split_dataframe

cfg = load_config()
import keras
from src.data import load_datasets
from src.evaluate import (predict_test, compute_metrics, per_class_report,
                        plot_confusion_matrix, top_errors, plot_misclassified)
```

```python
_, _, test_ds, test_df, classes = load_datasets(cfg)
model = keras.models.load_model(os.path.join(cfg["model_dir"], "modele_plantes.keras"))
y_true, y_pred = predict_test(model, test_ds, test_df)

acc, f1 = compute_metrics(y_true, y_pred)
print(f"Accuracy : {acc:.4f}")
print(f"F1 macro : {f1:.4f}")
```

## Résultats par classe

```python
print(per_class_report(y_true, y_pred, classes))
```

```python
plot_confusion_matrix(y_true, y_pred, classes)
plt.show()
```

```python
n_err, paires = top_errors(y_true, y_pred, classes)
print(n_err, "erreurs sur", len(y_true), "images de test\n")
paires
```
